```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
df = pd.read_csv("df = pd.read_csv(r"C:\Users\Bomathi\Downloads\archive (6)\ObesityDataSet.csv.csv")")

# First 64 students
df = df.head(64)

# Dataset Preview
print("Dataset Shape:", df.shape)
print(df.head())

# -------------------------------
# Histogram
# -------------------------------
plt.figure(figsize=(8,5))
sns.histplot(df['Weight'], bins=10, kde=True)
plt.title("Weight Distribution")
plt.xlabel("Weight")
plt.ylabel("Frequency")
plt.show()

# -------------------------------
# Violin Plot
# -------------------------------
plt.figure(figsize=(12,5))
sns.violinplot(x='Obesity', y='Weight', data=df)
plt.title("Weight Distribution by Obesity Category")
plt.xticks(rotation=45)
plt.show()

# -------------------------------
# Heatmap
# -------------------------------
plt.figure(figsize=(6,4))
corr = df[['Age','Height','Weight']].corr()

sns.heatmap(corr,
            annot=True,
            cmap='coolwarm')

plt.title("Correlation Heatmap")
plt.show()
```
